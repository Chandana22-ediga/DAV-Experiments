```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "student":["S1","S2","S3","S4","S5","S6","S7","S8","S9","S10"],
    "Study_hours":[2,3,4,5,6,7,8,9,4,5],
    "attendance":[65,70,72,75,78,80,82,85,68,74],
    "assignment_marks":[12,14,15,17,18,19,20,22,14,16],
    "internal_marks":[18,20,22,24,25,27,28,30,21,23],
    "final_marks":[45,50,55,60,65,70,75,80,52,58]
}
df = pd.DataFrame(data)
print(df)
```

```python
plt.hist(df["final_marks"],bins=5)
plt.xlabel("final_marks")
plt.ylabel("Number of students")
plt.title("distribution of final_marks")
plt.show()
```

```python
plt.hist(df["assignment_marks"],bins=5)
plt.xlabel("assignment_marks")
plt.ylabel("Number of students")
plt.title("distribution of assignment_marks")
plt.show()
```

```python
plt.hist(df["internal_marks"],bins=5)
plt.xlabel("internal_marks")
plt.ylabel("Number of students")
plt.title("distribution of internal_marks")
plt.show()
```

```python
plt.boxplot(df["final_marks"])
plt.xlabel("final_marks")
plt.ylabel("marks")
plt.title("box plot of final marks")
plt.show()
```

```python
plt.boxplot(df["assignment_marks"])
plt.xlabel("assignment_marks")
plt.ylabel("marks")
plt.title("box plot of assignment marks")
plt.show()
```

```python
plt.boxplot(df["internal_marks"])
plt.xlabel("internal_marks")
plt.ylabel("marks")
plt.title("box plot of internal marks")
plt.show()
```

```python
plt.boxplot([
    df["Study_hours"],
    df["attendance"],
    df["assignment_marks"],
    df["internal_marks"],
     df["final_marks"]
])
plt.xticks(
    [1,2,3,4,5],
    ["study hours","attendance","assignment ","internal","final"]
)
plt.ylabel("marks")
plt.title("box plot student performance")
plt.show()
```

```python
import seaborn as sns
sns.pairplot(
    df[
        ["Study_hours","attendance","assignment_marks","internal_marks","final_marks"]
    ]
)
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True)
plt.title("correlation matrix")
plt.show()
```
